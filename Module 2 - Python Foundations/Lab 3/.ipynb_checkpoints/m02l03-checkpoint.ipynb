```python
# Used Gemini to generate this list, still don't have textbook
 
# Freezing: 18, 32
# Cold: 45, 50
# Mild: 65, 70
# Warm: 82, 85
# Hot: 98
temperatures = [18, 32, 45, 50, 65, 70, 82, 85, 98]

for t in temperatures:
    if t < 45:
        type = "freezing"
    elif t < 65:
        type = "cold"
    elif t < 82:
        type = "mild"
    elif t < 98:
        type = "warm"
    else: type = "hot"
    print(f"{t} degrees is {type}")
```
