```python
# Step 2: Count the warm days

temps = [61, 65, 72, 58, 70, 74, 68]
count = 0
for temp in temps:
    if temp > 65:
        count += 1
print(count)
```

```python
# Step 3: Write a temperature converter

def to_celsius(f):
    return round((f - 32) * 5/9, 1)

print(to_celsius(temps[0]))
```
