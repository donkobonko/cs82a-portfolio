```python
# Don't really have the textbook so I'm freestyling here
# github repo: https://github.com/donkobonko/cs82a-portfolio

# For loop version
def multiply_for(a, b):
    result = 0
    for i in range(b):
        result += a
    return result

print(multiply_for(7, 6)) 
print(multiply_for(6, 7))
```

```python
# While loop version
def multiply_while(a, b):
    result = 0
    i = 0
    while i < b:
        result += a
        i += 1 # this is the termination condition, otherwise i will always just be stuck at 0 and the loop runs forever
    return result

print(multiply_while(7, 6))
print(multiply_while(6, 7))
```
