```python
# github repo: https://github.com/donkobonko/cs82a-portfolio

import pandas as pd

df = pd.read_csv("../Module 3 - Data Types, Storage, and Cleaning/Lab 1/sales_clean.csv")
print(df.shape)
```

```python
print("Price: ")
print(df["price"].mean())
print(df["price"].median())
print(df["price"].std())
print("\nQuantity: ")
print(df["qty"].mean())
print(df["qty"].median())
print(df["qty"].std())
```

> For the price, I'd report the median, as it seems like the mean is skewed high. The median would be more accurate in this case.

```python
df["price"].plot(kind="hist", bins=20)
```

> The histogram is skewed-right.

```python
new_df = df.copy() # Just to be non-destructive
new_df["revenue"] = new_df["price"] * new_df["qty"]
print(new_df.head())
```

```python
display(new_df.corr(numeric_only=True))
```

> price vs. revenue does in fact have the closest correlation.

```python
new_df.plot.scatter(x="price", y="revenue")
```

> Price and revenue have a strong positive correlation (r is around 0.82), and it can be seen on the scatter plot by the positive trend (higher prices result in higher revenues, as revenue is price * qty). There are only three outliers to look out for, however those three have already been flagged for a chance at being refunds (negative quantities).
