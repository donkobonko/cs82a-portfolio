plausibility rule  
I believe a reasonable range for the length should be from 900 to 3000. above 3000 seems to be getting blown out of proportion, especially if its a change more than 500-700, and anything under 900 might be starting to get too short and be of concern.

````python
import pandas as pd

df = pd.read_csv("bridges.csv")

print(df.head(10))
print(df.describe())
print(df.shape)
````

````python
print(df.dtypes)
print(df.isna().sum())
````

````python
flagged_lengths = df[(df["length"] < 900) | (df["length"] > 3000)]
print(flagged_lengths.sort_values(by="length", ascending=False))
````

````python
df_clean = df.copy()
df_clean["flagged-lengths"] = (df["length"] < 900) | (df["length"] > 3000)
print(df_clean[df_clean["flagged-lengths"]].sort_values(by="length", ascending=False))
````
